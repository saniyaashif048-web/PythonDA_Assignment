```python
import numpy as np


temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 Temperatures (C):", temperatures_w1)


print("\n Array Inspection ")
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)


print("\nArray Operations")
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_f)


print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", round(np.mean(temperatures_w1), 2))


print("\nArray Slicing ")
first_three_days = temperatures_w1[:3]
print("First three days:", first_three_days)

weekend_days = temperatures_w1[-2:]
print("Weekend (Last two days):", weekend_days)

middle_three_days = temperatures_w1[2:5]
print("Middle three days:", middle_three_days)
```

```python

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])


print("2D Array Inspection ")
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)

print("\n2D Array Slicing")

print("Week 1 temperatures:", temperatures[0, :])
print("Week 2 temperatures:", temperatures[1, :])

weekends = temperatures[:, -2:]
print("Weekends (both weeks):\n", weekends)
```

```python
import pandas as pd
data = [95, 92, 89, 85, 80]
indices = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
marks = pd.Series(data, index=indices)
print("--- Initial Pandas Series ---")
print(marks)


print("\n--- Indexing & Slicing ---")

print("Mark of 1st rank (position 0):", marks.iloc[0])

print("Top 3 ranks:\n", marks.loc['Rank1':'Rank3'])

print("Mark of 3rd rank (position 2):", marks.iloc[2])

mask = marks > 90
print("Ranks with marks > 90:\n", marks[mask])


print("\n--- Series Manipulation ---")

marks.loc['Rank1'] = 100


marks = marks.drop('Rank5')


cgpa = marks / 10

print("Final Modified Series:\n", marks)
print("\nComputed CGPA:\n", cgpa)
```

```python
import pandas as pd
data = {
    'TransactionID':[101, 102, 103, 104, 105, 106, 107, 108, 109, 110 ],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                       'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)

print("Data Exploration")
print("DataFrame Shape:", transactions.shape)
print("\nDataFrame Head:\n", transactions.head())
print("\nDataFrame Tail:\n", transactions.tail())
print("\nBasic Info:")
print(transactions.info())


print("\nSpecific Columns:\n", transactions[['ProductCategory', 'Amount']])


print("\nLast 3 Columns:\n", transactions.iloc[:, -3:])


filtered_df = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print("\nFiltered (Region=='North' & Amount>200):\n", filtered_df)


print("\nProduct Category Counts:\n", transactions['ProductCategory'].value_counts())


print("\nUnique Regions:", transactions['Region'].unique())


mean_per_region = transactions.groupby('Region')['Amount'].mean()
print("\nMean Amount per Region:\n", mean_per_region)


print("\n--- DataFrame Manipulation ---")

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

transactions['Discount'] = transactions['Amount'] * 0.10
print("Added Discount column:\n", transactions.head(3))


transactions = transactions[transactions['TransactionID'] != 109]


transactions = transactions.drop(columns=['Discount'])
print("\nFinal Transformed DataFrame:\n", transactions)
```
